```python
%python
# Databricks notebook source
from pyspark.sql.functions import col, to_date, row_number
from pyspark.sql.window import Window

# Step 1: Ensure silver schema exists in default catalog
spark.sql("CREATE CATALOG IF NOT EXISTS swap01")
spark.sql("CREATE SCHEMA IF NOT EXISTS swap01.silver_schema")

# Step 2: Load Bronze orders
bronze_orders = spark.table("swap01.bronze_schema.orders_bronze")

# Step 3: Rename and cast columns
orders_cleaned = (
    bronze_orders
    .select(
        col("order_id").alias("order_id_silver"),
    )
)

# Step 4: Write to Silver layer
orders_cleaned.write.format("delta").mode("overwrite").saveAsTable("swap01.silver_schema.orders_cleaned_silver")
print("✅ Silver cleaned complete: training_centralindia_lakeflowjobs_dbws.silver_schema.orders_cleaned_silver")
```
